##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
def porcentaje_A (cadena_texto):
    countA = cadena_texto.count('A')
    longitud = len(cadena_texto)
    porcentaje = countA / longitud
    return round(porcentaje,2)
```

```python
valor = porcentaje_A(secuencia)
valor
```

##### ¿Y que devuelva el número de decimales que yo quiero?

```python
def porcentaje_A (cadena_texto,decimales):
    countA = cadena_texto.count('A')
    longitud = len(cadena_texto)
    porcentaje = countA / longitud
    return round(porcentaje,decimales)
```

```python
valor = porcentaje_A(secuencia,1)
valor
```

##### ¿Sabes poner un valor por defecto en un argumento?

```python
def porcentaje_A (cadena_texto,decimales=2):
    countA = cadena_texto.count('A')
    longitud = len(cadena_texto)
    porcentaje = countA / longitud
    return round(porcentaje,decimales)
```

```python
valor = porcentaje_A(secuencia)
valor
```

##### ¿Qué pasa si no ponemos un argumento? ¿Sabrías hacer una función encapsulada?

```python
#funcion encapsulada es una función sin argumentos
def porcentaje_A ():
    countA = cadena_texto.count('A')
    longitud = len(cadena_texto)
    porcentaje = countA / longitud
    return round(porcentaje,decimales)
```

#### 5.- De la lista de accessions que se proporciona obtén lo siguiente:
1. Los que tengan el número 5
2. Los que tengan la letra 'd' o la 'e'
3. Los que tengan las letras 'd' y 'e' en ese orden
4. Los que tengan las letras 'd' y 'e' en ese orden con una letra cualquiera en medio
5. Los que tengan ambas letras 'd' y 'e' en cualquier orden
6. Los que empiezan por 'x' o por 'y'
7. Los que empiezan por 'x' o 'y' y terminan por 'e'
8. Los que tienen tres o más dígitos seguidos
9. Los que terminan con 'd' seguidos de 'a' o 'r' o 'p'

```python
import re
```

```python
accs = ['xkn59438', 'yhdck2', 'eihd39d9', 'chdsye847', 'hedle3455', 'xjhd53e', '45da', 'de37dp']
```

```python
#1:
for cosa in accs:
    if re.search('5',cosa):
        print(cosa)
```

```python
#2
for cosa in accs:
    if re.search('d', cosa) or re.search('e',cosa):
        print(cosa)
```

```python
#2
for cosa in accs:
    if re.search('d|e',cosa):
        print(cosa)
```

```python
#2
for cosa in accs:
    if re.search('[de]',cosa):
        print(cosa)
```

```python
#3
for cosa in accs:
    if re.search('de', cosa):
        print(cosa)
```

```python
#4
for cosa in accs:
    if re.search('d.*e', cosa):
        print(cosa)
```

```python
#5
for cosa in accs:
    if re.search('d', cosa) and re.search('e',cosa):
        print(cosa)
```

```python
#6
for cosa in accs:
    if re.search('^[xy]', cosa):
        print(cosa)
```

```python
#6
for cosa in accs:
    if re.search('^x', cosa) or re.search('^y',cosa):
        print(cosa)
```

```python
#7
for cosa in accs:
    if re.search('^[xy]', cosa) and re.search('e$',cosa):
        print(cosa)
```

```python
#8
for cosa in accs:
    if re.search('\d{3}', cosa):
        print(cosa)
```

```python
#8
for cosa in accs:
    if re.search('[0-9]{3}', cosa):
        print(cosa)
```

```python
#9
for cosa in accs:
    if re.search('d[arp]$', cosa):
        print(cosa)
```
